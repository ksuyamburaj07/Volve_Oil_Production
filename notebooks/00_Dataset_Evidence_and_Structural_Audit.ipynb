## 9. Daily-to-Monthly Reconciliation

The daily dataset is independently aggregated by well, year, and month.

These calculated monthly totals are then compared with the supplied Monthly Production Data sheet to verify that the daily data has been interpreted correctly.

```python
daily_for_monthly = daily_df.copy()

daily_for_monthly["Year"] = (
    daily_for_monthly["DATEPRD"].dt.year
)

daily_for_monthly["Month"] = (
    daily_for_monthly["DATEPRD"].dt.month
)


def sum_preserve_missing(series):
    return series.sum(min_count=1)


daily_monthly_agg = (
    daily_for_monthly
    .groupby(
        [
            "NPD_WELL_BORE_NAME",
            "NPD_WELL_BORE_CODE",
            "Year",
            "Month"
        ],
        as_index=False
    )
    .agg(
        daily_rows=(
            "DATEPRD",
            "size"
        ),

        flow_kinds=(
            "FLOW_KIND",
            lambda series:
                ", ".join(
                    sorted(
                        series.dropna().unique()
                    )
                )
        ),

        daily_on_stream=(
            "ON_STREAM_HRS",
            sum_preserve_missing
        ),

        daily_oil=(
            "BORE_OIL_VOL",
            sum_preserve_missing
        ),

        daily_gas=(
            "BORE_GAS_VOL",
            sum_preserve_missing
        ),

        daily_water=(
            "BORE_WAT_VOL",
            sum_preserve_missing
        ),

        daily_wi=(
            "BORE_WI_VOL",
            sum_preserve_missing
        )
    )
)

print(
    "Daily data aggregated to monthly rows:",
    len(daily_monthly_agg)
)

display(daily_monthly_agg.head(10))
mixed_role_months = daily_monthly_agg[
    daily_monthly_agg[
        "flow_kinds"
    ].str.contains(",")
]

print(
    "\nMonths containing more than one FLOW_KIND:",
    len(mixed_role_months)
)

display(mixed_role_months)
```

```python
monthly_for_compare = (
    monthly_clean
    .rename(
        columns={
            "Wellbore name":
                "NPD_WELL_BORE_NAME",

            "NPDCode":
                "NPD_WELL_BORE_CODE",

            "On Stream":
                "monthly_on_stream",

            "Oil":
                "monthly_oil",

            "Gas":
                "monthly_gas",

            "Water":
                "monthly_water",

            "WI":
                "monthly_wi"
        }
    )
    .copy()
)

comparison = daily_monthly_agg.merge(
    monthly_for_compare[
        [
            "NPD_WELL_BORE_NAME",
            "NPD_WELL_BORE_CODE",
            "Year",
            "Month",
            "monthly_on_stream",
            "monthly_oil",
            "monthly_gas",
            "monthly_water",
            "monthly_wi"
        ]
    ],

    on=[
        "NPD_WELL_BORE_NAME",
        "NPD_WELL_BORE_CODE",
        "Year",
        "Month"
    ],

    how="outer",
    validate="one_to_one",
    indicator=True
)

print("DAILY ↔ MONTHLY ROW ALIGNMENT")
print("-" * 60)

print(comparison["_merge"].value_counts())

display(comparison.head())
```

### 9.1 Numerical Reconciliation

For jointly observed values, daily-derived monthly totals are compared numerically with the supplied monthly totals using a small floating-point tolerance.

Missing-versus-zero representation differences are reported separately rather than automatically changed.

```python
import numpy as np


metric_pairs = {
    "on_stream": (
        "daily_on_stream",
        "monthly_on_stream"
    ),

    "oil": (
        "daily_oil",
        "monthly_oil"
    ),

    "gas": (
        "daily_gas",
        "monthly_gas"
    ),

    "water": (
        "daily_water",
        "monthly_water"
    ),

    "wi": (
        "daily_wi",
        "monthly_wi"
    )
}


reconciliation_rows = []
representation_frames = []


for metric, columns in metric_pairs.items():

    daily_column, monthly_column = columns

    both_present = (
        comparison[daily_column].notna()
        &
        comparison[monthly_column].notna()
    )

    one_side_missing = (
        comparison[daily_column].isna()
        ^
        comparison[monthly_column].isna()
    )

    daily_values = comparison.loc[
        both_present,
        daily_column
    ]

    monthly_values = comparison.loc[
        both_present,
        monthly_column
    ]

    absolute_difference = (
        daily_values - monthly_values
    ).abs()

    within_tolerance = np.isclose(
        daily_values,
        monthly_values,
        rtol=1e-9,
        atol=1e-6
    )

    reconciliation_rows.append({
        "metric": metric,

        "both_present":
            int(both_present.sum()),

        "within_tolerance":
            int(within_tolerance.sum()),

        "one_side_missing":
            int(one_side_missing.sum()),

        "maximum_absolute_difference":
            (
                absolute_difference.max()
                if len(absolute_difference) > 0
                else np.nan
            )
    })

    if one_side_missing.any():

        representation_difference = (
            comparison.loc[
                one_side_missing,
                [
                    "NPD_WELL_BORE_NAME",
                    "Year",
                    "Month",
                    daily_column,
                    monthly_column
                ]
            ]
            .copy()
        )

        representation_difference.insert(
            0,
            "metric",
            metric
        )

        representation_frames.append(
            representation_difference
        )


reconciliation_summary = pd.DataFrame(
    reconciliation_rows
)

print("NUMERICAL RECONCILIATION")
print("-" * 60)

display(reconciliation_summary)


if representation_frames:

    representation_differences = pd.concat(
        representation_frames,
        ignore_index=True
    )

    print(
        "\nMissing / zero representation differences:"
    )

    display(representation_differences)

else:

    representation_differences = (
        pd.DataFrame()
    )

    print(
        "\nNo one-sided missing-value "
        "differences found."
    )
```

## 10. Refined Active-Day Well Profiles

Recorded-day averages include shutdown days, while active-day averages use only records where ON_STREAM_HRS is greater than zero.

Both measures are retained because they describe different aspects of well performance.

```python
active_production_df = production_df[
    production_df["ON_STREAM_HRS"] > 0
].copy()

active_injection_df = injection_df[
    injection_df["ON_STREAM_HRS"] > 0
].copy()


producer_profile_refined = (
    production_df
    .groupby("NPD_WELL_BORE_NAME")
    .agg(
        total_records=(
            "DATEPRD",
            "size"
        ),

        active_days=(
            "ON_STREAM_HRS",
            lambda series:
                (series > 0).sum()
        ),

        inactive_days=(
            "ON_STREAM_HRS",
            lambda series:
                (series == 0).sum()
        ),

        total_oil=(
            "BORE_OIL_VOL",
            "sum"
        ),

        average_recorded_day_oil=(
            "BORE_OIL_VOL",
            "mean"
        ),

        maximum_daily_oil=(
            "BORE_OIL_VOL",
            "max"
        )
    )
    .reset_index()
)


producer_active_average = (
    active_production_df
    .groupby("NPD_WELL_BORE_NAME")[
        "BORE_OIL_VOL"
    ]
    .mean()
    .rename("average_active_day_oil")
    .reset_index()
)


producer_profile_refined = (
    producer_profile_refined
    .merge(
        producer_active_average,
        on="NPD_WELL_BORE_NAME",
        how="left"
    )
)


injector_profile_refined = (
    injection_df
    .groupby("NPD_WELL_BORE_NAME")
    .agg(
        total_records=(
            "DATEPRD",
            "size"
        ),

        active_days=(
            "ON_STREAM_HRS",
            lambda series:
                (series > 0).sum()
        ),

        inactive_days=(
            "ON_STREAM_HRS",
            lambda series:
                (series == 0).sum()
        ),

        unknown_activity_days=(
            "ON_STREAM_HRS",
            lambda series:
                series.isna().sum()
        ),

        total_water_injected=(
            "BORE_WI_VOL",
            "sum"
        ),

        average_available_day_injection=(
            "BORE_WI_VOL",
            "mean"
        ),

        maximum_daily_injection=(
            "BORE_WI_VOL",
            "max"
        )
    )
    .reset_index()
)


injector_active_average = (
    active_injection_df
    .groupby("NPD_WELL_BORE_NAME")[
        "BORE_WI_VOL"
    ]
    .mean()
    .rename(
        "average_active_day_injection"
    )
    .reset_index()
)


injector_profile_refined = (
    injector_profile_refined
    .merge(
        injector_active_average,
        on="NPD_WELL_BORE_NAME",
        how="left"
    )
)


print("REFINED PRODUCER PROFILE")
display(producer_profile_refined)

print("\nREFINED INJECTOR PROFILE")
display(injector_profile_refined)
```

## 11. Sensor Availability During Active Production

Sensor completeness is recalculated using only active production records.

This provides a more relevant view of which operational measurements may realistically be available for later modelling.

```python
sensor_availability_active = (
    active_production_df
    .groupby(
        "NPD_WELL_BORE_NAME"
    )[sensor_columns]
    .apply(
        lambda group:
            (
                group.notna().mean()
                * 100
            ).round(2)
    )
)

print(
    "SENSOR AVAILABILITY "
    "DURING ACTIVE PRODUCTION (%)"
)

display(sensor_availability_active)
```
